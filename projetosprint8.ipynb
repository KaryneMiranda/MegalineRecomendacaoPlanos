```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import RandomForestClassifier 
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold, train_test_split
from sklearn.metrics import accuracy_score, classification_report

df = pd.read_csv("/datasets/users_behavior.csv")
display(df.head())
display(df.info())
df.describe()
print()
print(df.isnull().sum())
```

```python
#Definindo as variáveis
x = df.drop('is_ultra', axis=1)
y = df['is_ultra']
```

```python
#Primeiro divide treino e teste
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
#Primeiro divide treino e validação
x_train, x_valid, y_train, y_valid = train_test_split(
    x_train,
    y_train,
    test_size=0.25,
    random_state=42
)
```

```python
print(x_train.shape)
print(x_valid.shape)
print(x_test.shape)
print()
print(len(x_train))
print(len(x_valid))
print(len(x_test))
```

```python
## Árvore de Decisão
arvore = DecisionTreeClassifier(random_state = 42)
arvore.fit(x_train, y_train)
pred_valid = arvore.predict(x_valid)
print(f"Árvore: {accuracy_score(y_valid, pred_valid):.2f}")
```

```python
parametros_arvore = {
    'max_depth': [3, 5, 7, 10, 15, None],
    'min_samples_split': [2, 5, 10, 20],
    'min_samples_leaf': [1, 2, 5, 10],
    'criterion': ['gini', 'entropy']
}

busca_arvore = RandomizedSearchCV(
    DecisionTreeClassifier(random_state=42),
    parametros_arvore,
    n_iter=20,
    cv=5,
    random_state=42,
    n_jobs=-1
)

busca_arvore.fit(x_train, y_train)
print("=== MELHOR ÁRVORE ===")
print(f"Parâmetros: {busca_arvore.best_params_}")
print(f"Acurácia validação: {busca_arvore.best_score_:.4f}")

# Testar no conjunto de validação manual
melhor_arvore = busca_arvore.best_estimator_
acc_valid = accuracy_score(y_valid, melhor_arvore.predict(x_valid))
print(f"Acurácia validação (manual): {acc_valid:.4f}")

print()
```

```python
## Floresta Aleatória
floresta = RandomForestClassifier(random_state = 42)
floresta.fit(x_train, y_train)
pred_valid = floresta.predict(x_valid)
print(f"Floresta: {accuracy_score(y_valid, pred_valid):.2f}")
```

```python
parametros_floresta = {
    'n_estimators': [50, 100, 150, 200, 300],
    'max_depth': [3, 5, 7, 10, 15, None],
    'min_samples_split': [2, 5, 10, 20],
    'min_samples_leaf': [1, 2, 5, 10]
}

busca_floresta = RandomizedSearchCV(
    RandomForestClassifier(random_state=42),
    parametros_floresta,
    n_iter=30,
    cv=5,
    random_state=42,
    n_jobs=-1
)

busca_floresta.fit(x_train, y_train)
print("=== MELHOR RANDOM FOREST ===")
print(f"Parâmetros: {busca_floresta.best_params_}")
print(f"Acurácia validação: {busca_floresta.best_score_:.4f}")

melhor_floresta = busca_floresta.best_estimator_
acc_valid = accuracy_score(y_valid, melhor_floresta.predict(x_valid))
print(f"Acurácia validação (manual): {acc_valid:.4f}")

print()
```

```python
## Regressão Logística
logistica = LogisticRegression(random_state = 42)
logistica.fit(x_train, y_train)
pred_valid = logistica.predict(x_valid)
print(f"Regressão Logística: {accuracy_score(y_valid,pred_valid):.2f}")
```

```python
parametros_log = {
    'C': [0.01, 0.1, 1, 10, 100],
    'solver': ['liblinear', 'lbfgs'],
    'max_iter': [500, 1000, 2000]
}

busca_log = RandomizedSearchCV(
    LogisticRegression(random_state=42),
    parametros_log,
    n_iter=15,
    cv=5,
    random_state=42,
    n_jobs=-1
)

busca_log.fit(x_train, y_train)
print("=== MELHOR REGRESSÃO LOGÍSTICA ===")
print(f"Parâmetros: {busca_log.best_params_}")
print(f"Acurácia validação: {busca_log.best_score_:.4f}")
```

```python
## Verificando a qualidade do modelo Random Forest com o conjunto de teste
modelo_final = RandomForestClassifier(n_estimators = 100, max_depth = 7, random_state = 42)
modelo_final.fit(x_train, y_train)

# Previsão do teste
y_pred_test = modelo_final.predict(x_test)

# Avaliar
print("AVALIAÇÃO NO CONJUNTO DE TESTE")
print(f"Acurácia: {accuracy_score(y_test, y_pred_test):.2f}")
```

```python

# Comparar com validação (para ver se não houve sobreajuste)
y_pred_valid = modelo_final.predict(x_valid)

acuracia_valid = accuracy_score(y_valid, y_pred_valid)
acuracia_teste = accuracy_score(y_test, y_pred_test)

print(f"\nAcurácia na validação: {acuracia_valid:.2f}")
print(f"Acurácia no teste:     {acuracia_teste:.2f}")
print(f"Diferença: {abs(acuracia_valid - acuracia_teste):.2f}")

# Relatório Detalhado
print('\n Relatório \n')
print(classification_report(y_test, y_pred_test))
```
